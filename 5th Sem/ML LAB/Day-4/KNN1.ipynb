```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
from collections import Counter
```

```python
data = pd.read_csv("KNN.csv")
data
```

```python
# X_Training Data
X_Train = data.iloc[:, 1:-1].values
X_Train
```

```python
# Y_Train
Y_Train = data.iloc[:, -1].values
Y_Train
```

```python
# Defining Euclidean Distance

def Euclidean_Distance(p1,p2):
    return math.sqrt(sum((a-b) ** 2 for a,b in zip(p1,p2)))
```

```python
# Defining KNN Algorithm

def KNN_Classify(X_Train, Y_Train, Test_Point, K):
    distances = []
    
    #Calculating Euclidean Distance
    for i in range(len(X_Train)):
        dist = Euclidean_Distance(Test_Point, X_Train[i])
        distances.append((dist, Y_Train[i]))

    print(distances)
    
    # Step 4: Sort distances and pick k nearest
    distances.sort(key=lambda x : x[0])
    k_nearest = [label for _, label in distances[:k]]

    # Step 5: Most Occured
    most_common = Counter(k_nearest).most_common(1)
    return most_common[0][0]
```

```python
# Example test point

Test_Point = [0, 0, 0]   # Test Point

k = 3 # K value

predicted_label = KNN_Classify(X_Train, Y_Train, Test_Point, k)   # Prdiction
```

```python
# Output

print(f"Test Point is : {Test_Point}")
print(f"The Predicted Value is: {predicted_label}")
```
