```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from itertools import combinations
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("BSS.csv")
data
```

```python
data['Sex'] = data['Sex'].map({"M":1, "F":0})
data
```

```python
X = data[['Age', 'Score', 'Sex']]
X
```

```python
Y = data['BP']
Y
```

```python
N = len(Y)
print("N:",N)
p = X.shape[1]
print("P:",p)
```

```python
X.columns
```

```python
res = []
for k in range(1, p+1):
    for features in combinations(X.columns, k):
        model = LinearRegression()
        model.fit(X[list(features)], Y)
        Y_Pred = model.predict(X[list(features)])
        rss = np.sum((Y - Y_Pred) ** 2)
        res.append([k, features, rss])

print(res)
```

```python
res = pd.DataFrame(res, columns=["K", "Best Predictors", "RSS"])
res
```

```python
best = res.loc[res["RSS"].idxmax()]
best["Best Predictors"]
```

```python
res = []
for k in range(1, p+1):
    for features in combinations(X.columns, k):
        model = LinearRegression()
        model.fit(X[list(features)], Y)
        Y_Pred = model.predict(X[list(features)])
        rss = np.sum((Y - Y_Pred) ** 2)
    res.append([k, features, rss])

print(res)
```

```python
res = pd.DataFrame(res, columns=["K", "Best Predictors", "RSS"])
res
```

```python
lowest_aic = float("inf")
for k, features, rss in res.values:
    aic = N*np.log(rss/N) + 2*k
    print(aic)
    if aic < lowest_aic:
        lowest_aic = aic
print(lowest_aic)
```

```python
lowest_bic = float("inf")
for k, features, rss in res.values:
    bic = N*np.log(rss/N) + k*np.log(N)
    print(bic)
    if bic < lowest_bic:
        lowest_bic = bic
print(lowest_bic)
```
