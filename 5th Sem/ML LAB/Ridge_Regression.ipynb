```python
import numpy as np
```

```python
X = np.array([[1,2],[2,4],[3,6]])
y = np.array([1,2,3])

lam = 0.5
```

```python
x_mean = np.mean(X, axis=0)
y_mean = np.mean(y)
```

```python
X = X-x_mean
y = y-y_mean
```

```python
X
```

```python
y
```

```python
I = np.eye(X.shape[1])
I
```

```python
beta_cap = np.dot(np.linalg.inv(np.dot(X.T,X) + lam * I), np.dot(X.T,y))
beta_cap
```

```python
beta_0_cap = y_mean - np.dot(x_mean, beta_cap)
beta_0_cap
```

```python
y_pred = beta_0_cap + np.dot(X, beta_cap)
y_pred
```

```python
RSS = np.sum((y - y_pred) ** 2)
```

```python
print(RSS)
```
