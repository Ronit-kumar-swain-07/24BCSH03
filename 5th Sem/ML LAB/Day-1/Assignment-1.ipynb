```python
# Importing the required libraries
import pandas as pd
import numpy as np


# Given dictionary data
data = {
    'birds': ['Cranes', 'Cranes', 'plovers', 'spoonbills', 'spoonbills','Cranes', 'plovers', 'Cranes', 'spoonbills', 'spoonbills'],
    'age': [3.5, 4, 1.5, np.nan, 6, 3, 5.5, np.nan, 8, 4],
    'visits': [2, 4, 3, 4, 3, 4, 2, 2, 3, 2],
    'priority': ['yes', 'yes', 'no', 'yes', 'no','no', 'no', 'yes', 'no', 'no']
}

# Given index labels
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
```

##### 1. Create a DataFrame birds from this dictionary data which has the index labels.

```python
# pd.DataFrame() converts dictionary data into a DataFrame
# index=labels gives custom row labels
birds = pd.DataFrame(data, index=labels)

birds
```

##### 2. Display a summary of the basic information about birds DataFrame and its data.

```python
# info() displays information such as: number of rows, columns, data types and memory usage
birds.info()
```

##### 3. Print the first 2 rows of the birds dataframe

```python
# head(n) returns the first n rows
birds.head(2)
```

##### 4. Print all the rows with only 'birds' and 'age' columns from the dataframe

```python
# [['birds', 'age']] selects only these two columns

birds[['birds', 'age']]
```

##### 5. select [2, 3, 7] rows and in columns ['birds', 'age', 'visits']

```python
# iloc[] selects rows using their position numbers
# 2, 3 and 7 represent the 3rd, 4th and 8th rows
birds.iloc[[2, 3, 7]][['birds', 'age', 'visits']]
```

##### 6. select the rows where the number of visits is less than 4

```python
# birds['visits'] < 4 creates a condition
# Only rows satisfying the condition are selected
birds[birds['visits'] < 4]
```

##### 7. select the rows with columns ['birds', 'visits'] where the age is missing i.e NaN

```python
# isna() checks for missing/NaN values
# loc[] selects rows and required columns
birds.loc[birds['age'].isna(), ['birds', 'visits']]
```

##### 8. Select the rows where the birds is a Cranes and the age is less than 4

```python
# == checks whether the bird is Cranes
# < 4 checks whether age is less than 4
birds[(birds['birds'] == 'Cranes') & (birds['age'] < 4)]
```

##### 9. Select the rows the age is between 2 and 4(inclusive)

```python
# between(2, 4) selects values from 2 to 4
birds[birds['age'].between(2, 4)]
```

##### 10. Find the total number of visits of the bird Cranes

```python
# First select only Cranes
# Then select their visits
# sum() calculates the total
total_visits = birds[birds['birds'] == 'Cranes']['visits'].sum()

print("Total visits of Cranes:", total_visits)
```

##### 11. Calculate the mean age for each different birds in dataframe.

```python
# groupby() groups rows according to bird type
# mean() calculates the average age of each group
mean_age = birds.groupby('birds')['age'].mean()

print(mean_age)
```

##### 12. Append a new row 'k' to dataframe with your choice of values for each column. Then delete that row to return the original DataFrame.


```python
# loc[] can be used to add a new row
birds.loc['k'] = ['eagle', 5, 3, 'yes']

print("After adding row k:")
print(birds)

# drop() removes the specified row
birds = birds.drop('k')

print("\nAfter deleting row k:")
print(birds)
```

##### 13. Find the number of each type of birds in dataframe (Counts)

```python
# value_counts() counts how many times each value occurs
bird_count = birds['birds'].value_counts()

print(bird_count)
```

##### 14. Sort dataframe (birds) first by the values in the 'age' in decending order, then by the value in the 'visits' column in ascending order.

```python
# sort_values() sorts the DataFrame
# age -> descending order (False)
# visits -> ascending order (True)
sorted_birds = birds.sort_values(by=['age', 'visits'],ascending=[False, True])
print(sorted_birds)
```

##### 15. Replace the priority column values with'yes' should be 1 and 'no' should be 0

```python
# map() replaces values according to the given dictionary
birds['priority'] = birds['priority'].map({'yes': 1,'no': 0})
birds
```

##### 16. In the 'birds' column, change the 'Cranes' entries to 'trumpeters'.

```python
birds['birds'] = birds['birds'].replace('Cranes','trumpeters')
birds
```
