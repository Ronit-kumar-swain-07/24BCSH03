```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
c1 = np.array([(4,1), (2,4), (2,3), (3,6), (4,4)])
c2 = np.array([(9,10), (6,8), (9,5), (8,7), (10,8)])
```

```python
c1
```

```python
c2
```

```python
m1 = np.mean(c1, axis = 0)
print(m1)
```

```python
m2 = np.mean(c2, axis = 0)
print(m2)
```

```python
d1 = c1-m1
d1
```

```python
s1 = d1.T @ d1
s1 /= 5
s1
```

```python
d2 = c2 - m2
d2
```

```python
s2 = d2.T @ d2
s2 /= 5
s2
```

```python
Sw = s1 + s2
Sw
```

```python
dm = np.array(m1 - m2)
dm
```

```python
dm = dm[:, np.newaxis]
dm
```

```python
Sb = np.dot(dm, dm.T)
Sb
```

```python
Swinv = np.linalg.inv(Sw)
Swinv
```

```python
res = Swinv * Sb
res
```

```python
I = np.eye(2)
I
```

```python
lam = np.linalg.eigvals(res)
lam
```
